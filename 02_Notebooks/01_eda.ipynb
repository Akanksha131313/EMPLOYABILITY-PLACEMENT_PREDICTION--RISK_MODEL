**7. Correlation Analysis**

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation Matrix")
plt.show()
```

-> **Impact:**

**8. Feature Impact on Employability**

**a) CGPA vs Employability -**

```python
sns.boxplot(x="Employable", y="CGPA", data=df)
plt.title("CGPA vs Employability")
plt.show()
```

**b) Technical Score vs Employability -**

```python
sns.boxplot(x="Employable", y="Technical_Score", data=df)
plt.title("Technical Score vs Employability")
plt.show()
```

**c) Backlogs vs Employability -**

```python
sns.boxplot(x="Employable", y="Backlogs", data=df)
plt.title("Backlogs vs Employability")
plt.show()
```

**Purpose:**

• Identify risk indicators

• Validate hypothesis: academic backlog reduces employability
